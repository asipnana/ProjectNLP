```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
baseline = pd.read_csv("../results/baseline_results.csv")
fasttext = pd.read_csv("../results/fasttext_results.csv")
indobert = pd.read_csv("../results/indobert_results.csv")
xlmr = pd.read_csv("../results/xlmr_results.csv")
```

```python
comparison = pd.concat(
    [baseline, fasttext, indobert, xlmr],
    ignore_index=True
)
```

```python
comparison = comparison.sort_values(
    by="Accuracy",
    ascending=False
).reset_index(drop=True)

comparison.insert(0, "Rank", range(1, len(comparison)+1))

comparison
```

```python
comparison.to_csv(
    "../results/model_comparison.csv",
    index=False
)
```

```python
best = comparison.iloc[0]

print("=" * 50)
print("BEST PERFORMING MODEL")
print("=" * 50)
print(f"Model      : {best['Model']}")
print(f"Accuracy   : {best['Accuracy']:.3f}")
print(f"Precision  : {best['Precision']:.3f}")
print(f"Recall     : {best['Recall']:.3f}")
print(f"F1-Score   : {best['F1-Score']:.3f}")
```
